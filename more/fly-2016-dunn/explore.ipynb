```python
def squeeze_rec(array):
	if type(array) is not np.ndarray:
		return array

	if array.shape[0] == 1:
		array = np.squeeze(array)

	for i in range(array.shape[0]):
		array[i] = squeeze_rec(array[i])

	return array
```
