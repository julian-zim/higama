```python
import numpy as np
import matplotlib.pyplot as plt

# Choose a radius
r = 5

# Create angle values
theta = np.linspace(0, 2 * np.pi, 400)

# Circle coordinates
x_circle = r * np.cos(theta)
y_circle = r * np.sin(theta)

# Horizontal line at y = 3/5 r
y_line = (3/5) * r
x_line = np.linspace(-(4/5) * r, (4/5) * r, 200)
y_line_array = np.full_like(x_line, y_line)

# Plot
plt.figure()
plt.plot(x_circle, y_circle)
plt.plot(x_line, y_line_array)

# Formatting
plt.gca().set_aspect('equal', 'box')
plt.axhline(0)
plt.axvline(0)
plt.title("Circle and Horizontal Line")
plt.show()
```
