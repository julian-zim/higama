```python
import matplotlib.pyplot as plt
import numpy as np
import math
import ipywidgets as widgets
from IPython.display import display


def sigmoid(x):
	return 1. / (1. + np.exp(-x))

def get_ds(x, x_max):
    switches_pct = (12, 30, 45, 50, 67, 81, 90, 95)
    thresholds = [x_max * s / 100.0 for s in switches_pct]
    flips = sum(x >= t for t in thresholds)  # number of thresholds passed
    return -1 if (flips % 2 == 0) else 1


def plot(clm_w, hgm_w1, hgm_w2):
	step = 0.01
	x_max = 8. * math.pi
	xs = np.arange(0, x_max, step)

	clm_ts = np.array([clm_w for x in xs])
	hgm_ts = np.array([hgm_w1 * sigmoid(get_ds(x, x_max) * hgm_w2) for x in xs])

	fig = plt.figure(figsize=(16, 4))
	ax = plt.axes()
	ax.set_xlim(0, x_max)
	ax.set_ylim(-2, 2)
	ax.plot(xs, clm_ts, alpha=0.75, color='green')
	ax.plot(xs, hgm_ts, alpha=0.75, color='blue')
	plt.show()


clm_w_slider = widgets.FloatSlider(min=0, max=2, step=0.01, value=1, description='clm w')

hgm_w1_slider = widgets.FloatSlider(min=0, max=2, step=0.01, value=2, description='hgm w 1')
hgm_w2_slider = widgets.FloatSlider(min=0, max=10, step=0.01, value=0, description='hgm w 2')

row0 = widgets.HBox([clm_w_slider])
row1 = widgets.HBox([hgm_w1_slider, hgm_w2_slider])

ui = widgets.VBox([row0, row1])

out = widgets.interactive_output(
	plot, {
		'clm_w': clm_w_slider,
		'hgm_w1': hgm_w1_slider, 'hgm_w2': hgm_w2_slider,
	}
)

display(ui, out)
```
