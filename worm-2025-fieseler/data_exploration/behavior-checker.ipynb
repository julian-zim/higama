```python
import sys
import os
import math

from matplotlib import pyplot as plt

import numpy as np


sys.path.append(os.path.join(os.getcwd(), '../import'))

from utils import read_utils
```

```python
path = os.path.join(read_utils.get_data_path(), 'worm-2025-fieseler', 'exported_data_')
filenames, continuous_behaviors, discrete_states, eigenworms, kymographs, traces = read_utils.get_data(path)
```

```python
n_plots = 9
dim = math.ceil(math.sqrt(n_plots))
fig, axes = plt.subplots(dim, dim, figsize=(15, 15))
for i in range(n_plots):
	axes[i // dim, i % dim].set_xticklabels([])
	axes[i // dim, i % dim].set_yticklabels([])
	axes[i // dim, i % dim].grid(False)

for behavior in continuous_behaviors:
	if behavior is not None:
		for i, key in enumerate(behavior):
			data = behavior[key]
			axes[i // dim, i % dim].grid(True)
			axes[i // dim, i % dim].plot(data)
			axes[i // dim, i % dim].set_title(key, fontsize=8)
```

```python
n_plots = 8
dim = math.ceil(math.sqrt(n_plots))
fig, axes = plt.subplots(dim, dim, figsize=(15, 15))
for i in range(n_plots):
	axes[i // dim, i % dim].set_xticklabels([])
	axes[i // dim, i % dim].set_yticklabels([])
	axes[i // dim, i % dim].grid(False)

for state in discrete_states:
	if state is not None:
		for i, key in enumerate(state):
			data = state[key]
			axes[i // dim, i % dim].grid(True)
			axes[i // dim, i % dim].plot(data)
			axes[i // dim, i % dim].set_title(key, fontsize=8)
```

```python
n_plots = 34
dim = math.ceil(math.sqrt(n_plots))
fig, axes = plt.subplots(dim, dim, figsize=(15, 15))
for i in range(n_plots):
	axes[i // dim, i % dim].set_xticklabels([])
	axes[i // dim, i % dim].set_yticklabels([])
	axes[i // dim, i % dim].grid(False)

for i, eigenworm in enumerate(eigenworms):
	if eigenworm is not None:
		axes[i // dim, i % dim].grid(True)
		axes[i // dim, i % dim].plot(eigenworm)
	axes[i // dim, i % dim].set_title(f'eigenworms {i}', fontsize=8)
```
