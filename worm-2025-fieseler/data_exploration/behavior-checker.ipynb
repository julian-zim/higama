```python
import sys
import os
import math

from matplotlib import pyplot as plt

import numpy as np


sys.path.append(os.path.join(os.getcwd(), '../import'))

from utils import read_utils
from utils import plot_utils
```

```python
path = os.path.join(read_utils.get_data_path(), 'worm-2025-fieseler', 'exported_data_')
filenames, continuous_behaviors, discrete_states, eigenworms, kymographs, traces = read_utils.get_data(path)
```

```python
top_space = 0
plots_per_row = 4
```

```python
plot_utils.subplot(continuous_behaviors, 9, plots_per_row, top_space)
```

```python
plot_utils.subplot(discrete_states, 8, plots_per_row, top_space)
```

```python
plot_utils.subplot(eigenworms, 5, plots_per_row, top_space)
```

```python
n_plots = 34
plots_per_row = 1

plots_per_col = math.ceil(n_plots / plots_per_row)
figheight = 2 * plots_per_col + top_space
fig, axes = plt.subplots(plots_per_col, plots_per_row, figsize=(16 * plots_per_row, figheight), squeeze=False)
for i in range(n_plots):
	axes[i // plots_per_row, i % plots_per_row].set_xticklabels([])
	axes[i // plots_per_row, i % plots_per_row].set_yticklabels([])
	axes[i // plots_per_row, i % plots_per_row].grid(False)

for i, kymograph in enumerate(kymographs):
	if kymograph is not None:
		axes[i // plots_per_row, i % plots_per_row].imshow(np.clip(kymograph, -0.01, 0.01).T, cmap='coolwarm')
	axes[i // plots_per_row, i % plots_per_row].set_title(f'kymograph {i}', fontsize=8)
```
