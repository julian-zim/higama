```python
import matplotlib.pyplot as plt
import numpy as np
import ipywidgets as widgets
from IPython.display import display


def lss(Xt, St):
	return St @ np.linalg.inv(St.T @ St) @ St.T @ Xt


def plot(w1, w2, x1_1, x1_2, x2_1, x2_2):
    Xt = np.array([
        [x1_1, x1_2],
        [x2_1, x2_2],
    ])

    Wt = np.array([w1, w2]).reshape(-1, 1)
    St = Xt @ Wt
    Xt_lss = lss(Xt, St)
    Xt_pinv = St @ np.linalg.pinv(Wt)

    fig = plt.figure(figsize=(8, 8))
    ax = plt.axes()
    ax.set_xlim(0, 8)
    ax.set_ylim(0, 8)

    ax.scatter(Xt_lss[:, 0], Xt_lss[:, 1], alpha=0.75, color='green')
    ax.plot([0, Xt_lss[1, 0] * 10], [0, Xt_lss[1, 1] * 10], alpha=0.75, color='green')

    ax.scatter(Xt_pinv[:, 0], Xt_pinv[:, 1], alpha=0.75, color='blue')
    ax.plot([0, Xt_pinv[1, 0] * 10], [0, Xt_pinv[1, 1] * 10], alpha=0.75, color='blue')

    ax.scatter(Xt[:, 0], Xt[:, 1], alpha=0.75, color='red')
    plt.show()


w1_slider = widgets.FloatLogSlider(base=10, min=-1, max=1, step=0.01, value=1, description='w 1')
w2_slider = widgets.FloatLogSlider(base=10, min=-1, max=1, step=0.01, value=1, description='w 1')

x1_1_slider = widgets.FloatSlider(min=0, max=8, step=0.1, value=4, description='x 1')
x1_2_slider = widgets.FloatSlider(min=0, max=8, step=0.1, value=2, description='y 1')
x2_1_slider = widgets.FloatSlider(min=0, max=8, step=0.1, value=5, description='x 2')
x2_2_slider = widgets.FloatSlider(min=0, max=8, step=0.1, value=1, description='y 2')

x_row1 = widgets.HBox([x1_1_slider, x1_2_slider])
x_row2 = widgets.HBox([x2_1_slider, x2_2_slider])

ui = widgets.VBox([w1_slider, w2_slider, x_row1, x_row2])

out = widgets.interactive_output(
    plot, {
        'w1': w1_slider, 'w2': w2_slider,
        'x1_1': x1_1_slider, 'x1_2': x1_2_slider,
        'x2_1': x2_1_slider, 'x2_2': x2_2_slider
    }
)

display(ui, out)
```
