```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
import nibabel as nib
```

```python
file = scipy.io.loadmat('6163622/spont_M150824_MP019_2016-04-05.mat')
print(file.keys())
```

```python
fig, axes = plt.subplots(nrows=7, ncols=4, figsize=(20, 20))
fig.subplots_adjust(hspace=1.5)
fig.subplots_adjust(wspace=0.5)


neuronal_traces = file['Fsp'].T

ax = axes[0, 0]
ax.imshow(neuronal_traces[:200, :100].T, aspect='auto', cmap='viridis', origin='lower')
ax.set_title('Neuronal activity :100/11983')

ax = axes[0, 1]
ax.plot(neuronal_traces[:200, 0])
ax.set_xlabel("Time (samping rate unknown)")
ax.set_title('Neuronal Trace 1/11983')

ax = axes[0, 2]
ax.plot(neuronal_traces[:200, 1])
ax.set_xlabel("Time (samping rate unknown)")
ax.set_title('Neuronal Trace 2/11983')

ax = axes[0, 3]
ax.plot(neuronal_traces[:200, 2])
ax.set_xlabel("Time (samping rate unknown)")
ax.set_title('Neuronal Trace 3/11983')


beh_face = file['beh']['face'][0][0]
beh_face_avg = beh_face['avgframe'][0][0]

ax = axes[1, 0]
ax.imshow(beh_face_avg)
ax.set_title('average face frame')


beh_face_motionmask = beh_face['motionMask'][0][0].T

ax = axes[1, 1]
ax.imshow(beh_face_motionmask[0].T)
ax.set_title('face roi importance pc 1/1000')
ax = axes[1, 2]
ax.imshow(beh_face_motionmask[1].T)
ax.set_title('face roi importance pc 2/1000')
ax = axes[1, 3]
ax.imshow(beh_face_motionmask[2].T)
ax.set_title('face roi importance pc 3/1000')


beh_face_motionsvd = beh_face['motionSVD'][0][0].T

ax = axes[2, 0]
ax.plot(beh_face_motionsvd[0, :1000])
ax.set_title('face roi pc 1/1000')

ax = axes[2, 1]
ax.plot(beh_face_motionsvd[1, :1000])
ax.set_title('face roi pc 2/1000')

ax = axes[2, 2]
ax.plot(beh_face_motionsvd[2, :1000])
ax.set_title('face roi pc 3/1000')

ax = axes[2, 3]
ax.plot(beh_face_motionsvd[3, :1000])
ax.set_title('face roi pc 4/1000')













beh_whisker = file['beh']['whisker'][0][0]


beh_whisker_motionmask = beh_whisker['motionMask'][0][0].T

ax = axes[3, 0]
ax.imshow(beh_whisker_motionmask[0].T)
ax.set_title('whisker roi importance pc 1/1000')
ax = axes[3, 1]
ax.imshow(beh_whisker_motionmask[0].T)
ax.set_title('whisker roi importance pc 2/1000')
ax = axes[3, 2]
ax.imshow(beh_whisker_motionmask[1].T)
ax.set_title('whisker roi importance pc 3/1000')
ax = axes[3, 3]
ax.imshow(beh_whisker_motionmask[2].T)
ax.set_title('whisker roi importance pc 4/1000')


beh_whisker_motionsvd = beh_whisker['motionSVD'][0][0].T

ax = axes[4, 0]
ax.plot(beh_whisker_motionsvd[0, :1000])
ax.set_title('whisker roi pc 1/1000')

ax = axes[4, 1]
ax.plot(beh_whisker_motionsvd[1, :1000])
ax.set_title('whisker roi pc 2/1000')

ax = axes[4, 2]
ax.plot(beh_whisker_motionsvd[2, :1000])
ax.set_title('whisker roi pc 3/1000')

ax = axes[4, 3]
ax.plot(beh_whisker_motionsvd[3, :1000])
ax.set_title('whisker roi pc 4/1000')







beh_eye = file['beh']['eye'][0][0]



beh_eye_motionsvd = beh_eye['motionSVD'][0][0].T

ax = axes[5, 0]
ax.plot(beh_eye_motionsvd[0, :1000])
ax.set_title('eye roi pc 1/1000')

ax = axes[5, 1]
ax.plot(beh_eye_motionsvd[1, :1000])
ax.set_title('eye roi pc 2/1000')

ax = axes[5, 2]
ax.plot(beh_eye_motionsvd[2, :1000])
ax.set_title('eye roi pc 3/1000')

ax = axes[5, 3]
ax.plot(beh_eye_motionsvd[3, :1000])
ax.set_title('eye roi pc 4/1000')











beh_pupil = file['beh']['pupil'][0][0]

beh_pupil_area = beh_pupil['area'][0][0]

ax = axes[6, 0]
ax.plot(beh_pupil_area[:1000])
ax.set_title('pupil aera')


beh_pupil_com = beh_pupil['com'][0][0]

ax = axes[6, 1]
ax.plot(beh_pupil_com[:1000])
ax.set_title('pupil center of mass x y')


runspeed = file['beh']['runSpeed'][0][0]

ax = axes[6, 2]
ax.plot(runspeed[:1000])
ax.set_title('running speed')
```
