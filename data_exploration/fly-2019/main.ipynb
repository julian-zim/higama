```python
import numpy as np
import nibabel as nib
import matplotlib.pyplot as plt
import scipy
```

```python
behavior_file = scipy.io.loadmat('GSB7/GSB7Xk.mat')
print(behavior_file.keys())
behavior_data = behavior_file['Xk']

'''plt.plot(data[:, 3], label='walk', color='green')
plt.plot(data[:, 4], label='groom', color='blue')
plt.plot(data[:, 5], label='rest', color='red')'''



plt.plot(behavior_data, label=['fwd speed', 'right turn', 'left turn', 'walk', 'groom', 'rest'])

plt.legend()
plt.show()
```

```python
fluorescence_image = nib.load('GSB7/GSB7_psfproj_dff_kf.nii')
fluorescence_image_data = fluorescence_image.get_fdata()


'''
Calculate the average of images and plot mean value over time.
'''
M = np.mean(np.mean(np.mean(fluorescence_image_data, 0), 0), 0)
plt.plot(M)
plt.show()


'''
Plot the middle voxel of each image over time.
'''
voxel = fluorescence_image_data[fluorescence_image_data.shape[0]//2, fluorescence_image_data.shape[1]//2, fluorescence_image_data.shape[2]//2, :]
plt.plot(voxel)
```

```python
components_nii = nib.load('GSB7/Components/GSB7_psfproj_dff_kf_232ICs_map.nii')
components_nii_data = components_nii.get_fdata()


'''
Calculate the average of images and plot mean value over time.
'''
M = np.mean(np.mean(np.mean(components_nii_data, 0), 0), 0)
plt.plot(M)
plt.show()


'''
Plot the middle voxel of each image over time.
'''
voxel = components_nii_data[components_nii_data.shape[0]//2, components_nii_data.shape[1]//2, components_nii_data.shape[2]//2, :]
plt.plot(voxel)
```
